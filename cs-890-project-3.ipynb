Building Topology Using Networkx

```python
# ================================
# STEP 1: FINAL HOST SET
# ================================

hosts = set(filtered_df['Src IP']).union(set(filtered_df['Dst IP']))

print("Final number of hosts:", len(hosts))
```

```python
import networkx as nx

G = nx.Graph()

# =========================
# STEP 1: DEFINE NODES
# =========================

hosts = list(set(filtered_df['Src IP']).union(set(filtered_df['Dst IP'])))
hosts = hosts[:20]  # keep 20 hosts

# 🔥 INCREASED SWITCHES (IMPORTANT FOR IQN ROUTING SPACE)
switches = [f"S{i}" for i in range(1, 16)]  # 15 switches

# Add nodes
for h in hosts:
    G.add_node(h, type="host")

for s in switches:
    G.add_node(s, type="switch")

print("Hosts:", len(hosts))
print("Switches:", len(switches))
```

```python
# =========================
# STEP 2: HOST → SWITCH CONNECTION
# =========================

import random

host_to_switch = {}

# shuffle switches for better distribution
random.shuffle(switches)

for i, host in enumerate(hosts):
    sw = switches[i % len(switches)]

    G.add_edge(host, sw, weight=1)
    host_to_switch[host] = sw

print("Host-switch links created")
```

```python
# =========================
# STEP 3: SWITCH → SWITCH (RING)
# =========================

import random

# =========================
# STEP 3: SWITCH → SWITCH (RING + EXTRA LINKS)
# =========================

# 1. Ring backbone
for i in range(len(switches)):
    G.add_edge(
        switches[i],
        switches[(i + 1) % len(switches)],
        weight=1
    )

# 2. Add extra random links (VERY IMPORTANT FOR IQN)
extra_links = len(switches) * 2  # adjustable complexity

for _ in range(extra_links):
    s1, s2 = random.sample(switches, 2)
    G.add_edge(s1, s2, weight=1)

print("Enhanced switch backbone created")
```

```python
# =========================
# STEP 4: TRAFFIC WEIGHTING
# =========================

edge_traffic = filtered_df.groupby(
    ['Src IP', 'Dst IP']
)['Total_Packets_sum'].sum()

for (src, dst), value in edge_traffic.items():
    if G.has_edge(src, dst):
        G[src][dst]['weight'] = value

print("Traffic weights applied")
```

```python
# Normalize weights so DRL training is stable:

weights = [d['weight'] for _, _, d in G.edges(data=True)]
max_w = max(weights)

for u, v, d in G.edges(data=True):
    d['weight'] = d['weight'] / max_w
```

```python
import matplotlib.pyplot as plt
import numpy as np

# =========================
# LAYOUT DESIGN
# =========================

pos = {}

# 🔷 Place switches in larger circle
angle_step = 2 * np.pi / len(switches)

for i, sw in enumerate(switches):
    angle = i * angle_step
    pos[sw] = (3 * np.cos(angle), 3 * np.sin(angle))  # bigger radius

# 🔷 Place hosts near their connected switch
for h in hosts:
    sw = list(G.neighbors(h))[0]

    offset_angle = np.random.uniform(0, 2*np.pi)
    radius = 0.8  # slightly bigger spread

    pos[h] = (
        pos[sw][0] + radius * np.cos(offset_angle),
        pos[sw][1] + radius * np.sin(offset_angle)
    )

# =========================
# DRAW GRAPH
# =========================

plt.figure(figsize=(12, 10))

# switches
nx.draw_networkx_nodes(
    G, pos,
    nodelist=switches,
    node_color="red",
    node_size=700,
    label="Switches"
)

# hosts
nx.draw_networkx_nodes(
    G, pos,
    nodelist=hosts,
    node_color="skyblue",
    node_size=250,
    label="Hosts"
)

# edges
nx.draw_networkx_edges(G, pos, width=1.0, alpha=0.6)

nx.draw_networkx_labels(G, pos, font_size=7)

plt.title("SDN Topology (20 Hosts + 15 Switch)")
plt.axis("off")
plt.legend()
plt.show()
```

```python
# ================================
# STEP 3: SEQUENCE CREATION
# ================================

import numpy as np

WINDOW_SIZE = 50   # past 50 minutes
HORIZON = 3        # predict next 3 minutes

def create_sequences(data_X, data_y):
    X, y = [], []

    for i in range(len(data_X) - WINDOW_SIZE - HORIZON):
        X.append(data_X[i:i+WINDOW_SIZE])
        y.append(data_y[i+WINDOW_SIZE:i+WINDOW_SIZE+HORIZON])

    return np.array(X), np.array(y)
```

```python
# ================================
# STEP 4: TCN MODEL (HIGH ACCURACY)
# ================================

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
from tcn import TCN

def build_tcn(input_shape, horizon):

    model = Sequential([
        TCN(
            nb_filters=128,
            kernel_size=3,
            dilations=[1, 2, 4, 8, 16],
            dropout_rate=0.05,
            return_sequences=False
        ),
        Dense(128, activation='relu'),
        Dense(64, activation='relu'),
        Dense(horizon)
    ])

    model.compile(
        optimizer='adam',
        loss='mse',
        metrics=['mae']
    )

    return model
```

```python
# ================================
# STEP 5: TRAIN + FULL ROLLING PREDICTION
# ================================

pair_predictions = {}

pairs = filtered_df.groupby(['Src IP', 'Dst IP'])

for (src, dst), group in pairs:

    group = group.sort_values('Minute')

    if len(group) < 50:
        continue

    print(f"{src} → {dst} | Raw rows: {len(group)}")

    X_data = group[selected_features].values
    y_data = group[target_col].values  # ❗ REAL SCALE (NO RESHAPE TO 0–1)

    X_seq, y_seq = create_sequences(X_data, y_data)

    if len(X_seq) < 30:
        continue

    print(f"Training for {src} → {dst} | Samples: {len(X_seq)}")

    model = build_tcn((WINDOW_SIZE, len(selected_features)), HORIZON)

    model.fit(
        X_seq, y_seq,
        epochs=40,
        batch_size=8,
        verbose=0
    )

    
    # ================================
    # FIXED ROLLING PREDICTION (AVERAGED)
    # ================================

    pred_sum = np.zeros(len(y_data))
    pred_count = np.zeros(len(y_data))

    for i in range(len(X_data) - WINDOW_SIZE - HORIZON):

        window = X_data[i:i+WINDOW_SIZE]
        window = window.reshape(1, WINDOW_SIZE, len(selected_features))

        pred = model.predict(window, verbose=0).flatten()

        for j in range(HORIZON):
            idx = i + WINDOW_SIZE + j
            pred_sum[idx] += pred[j]
            pred_count[idx] += 1

    # avoid divide by zero
    pred_series = np.divide(pred_sum, pred_count, where=pred_count > 0)
    pred_series[pred_count == 0] = np.nan
    # ================================
    # Rolling prediction
    # ================================
    # pred_series = np.full(len(y_data), np.nan)

    # for i in range(len(X_data) - WINDOW_SIZE - HORIZON):

    #     window = X_data[i:i+WINDOW_SIZE]
    #     window = window.reshape(1, WINDOW_SIZE, len(selected_features))

    #     pred = model.predict(window, verbose=0).flatten()

    #     pred_series[i+WINDOW_SIZE:i+WINDOW_SIZE+HORIZON] = pred

    pair_predictions[(src, dst)] = {
        "actual": y_data,
        "predicted": pred_series
    }
```

```python
class RoutingEnv:
    def __init__(self, G, max_traffic=500.0):
        self.G = G
        self.max_traffic = max_traffic

    def reset(self, src, dst, traffic):
        self.current = src
        self.dst = dst
        self.traffic = traffic
        self.path = [src]
        return self.current

    def step(self, action):
        next_node = action
    
        # 🚨 Prevent invalid moves
        if next_node not in self.G[self.current]:
            print(f"[ENV ERROR] Invalid transition: {self.current} → {next_node}")
            return self.current, -10, True
    
        prev_node = self.current   # ✅ FIX
        self.current = next_node
        self.path.append(next_node)
    
        # ✅ Correct edge access
        edge_weight = self.G[prev_node][next_node].get('weight', 1.0)
    
        scaled_traffic = max(0, self.traffic) / self.max_traffic
        reward = -edge_weight * scaled_traffic
    
        done = (self.current == self.dst)  # ✅ FIX (was wrong before)
    
        return self.current, reward, done
```

```python
def get_neighbors(G, node, dst):
    neighbors = list(G.neighbors(node))
    
    # If current node is a SWITCH → allow switches AND the destination host
    if G.nodes[node]['type'] == 'switch':
        neighbors = [n for n in neighbors if G.nodes[n]['type'] == 'switch' or n == dst]
    
    return neighbors
```

```python
# IQN Model
import torch
import torch.nn as nn

class IQN(nn.Module):
    def __init__(self, state_dim, action_dim):
        super(IQN, self).__init__()
        self.fc = nn.Sequential(
            nn.Linear(state_dim + 1, 128),
            nn.ReLU(),
            nn.Linear(128, action_dim)
        )

    def forward(self, state, tau):
        # state: (batch_size, state_dim)
        # tau: (batch_size, 1)
        x = torch.cat([state, tau], dim=1)
        return self.fc(x)
```

```python
# state representation
def encode_state(node, dst, traffic, G, link_loads=None):
    """
    node: current switch/host (string)
    dst: destination host (string)
    traffic: predicted traffic volume (float)
    G: networkx graph
    link_loads: optional dict mapping (u,v) -> current load (for dynamic routing)
    
    Returns a numpy array of meaningful features.
    """
    # 1. Distance (shortest path hops) to destination
    try:
        dist = nx.shortest_path_length(G, node, dst)
    except:
        dist = 10  # large penalty if unreachable
    
    # 2. Node degree (number of neighbours)
    degree = G.degree(node)
    
    # 3. Average link weight of outgoing edges (lower = less congested)
    out_weights = [G[node][nbr]['weight'] for nbr in G.neighbors(node)]
    avg_weight = np.mean(out_weights) if out_weights else 1.0
    
    # 4. Minimum weight among outgoing edges (best immediate next hop)
    min_weight = np.min(out_weights) if out_weights else 1.0
    
    # 5. Traffic volume (normalised)
    # traffic is already a float, keep as is
    
    # Combine into a fixed‑length vector
    node_type = 1 if G.nodes[node]['type'] == 'switch' else 0

    state_vec = np.array([
        dist,
        degree,
        avg_weight,
        min_weight,
        max(0, traffic),  # remove negative traffic
        node_type
    ], dtype=np.float32)
    return state_vec
```

```python
import torch
import torch.nn as nn
import torch.optim as optim
import random
import numpy as np
from collections import deque

# ---------- IQN Model (unchanged) ----------
class IQN(nn.Module):
    def __init__(self, state_dim, action_dim):
        super(IQN, self).__init__()
        self.state_dim = state_dim
        self.action_dim = action_dim
        self.fc = nn.Sequential(
            nn.Linear(state_dim + 1, 128),
            nn.ReLU(),
            nn.Linear(128, action_dim)
        )

    def forward(self, state, tau):
        batch_size = state.shape[0]
        num_quantiles = tau.shape[1]
        state_expanded = state.unsqueeze(1).expand(-1, num_quantiles, -1)
        tau_expanded = tau.unsqueeze(-1)
        x = torch.cat([state_expanded, tau_expanded], dim=-1)
        x = x.view(-1, self.state_dim + 1)
        q = self.fc(x)
        q = q.view(batch_size, num_quantiles, self.action_dim)
        return q

# ---------- Replay Buffer (unchanged) ----------
class ReplayBuffer:
    def __init__(self, capacity=10000):
        self.buffer = deque(maxlen=capacity)
    def push(self, state, action, reward, next_state, done):
        self.buffer.append((state, action, reward, next_state, done))
    def sample(self, batch_size):
        batch = random.sample(self.buffer, batch_size)
        state, action, reward, next_state, done = map(np.array, zip(*batch))
        return state, action, reward, next_state, done
    def __len__(self):
        return len(self.buffer)

# ---------- Quantile Huber Loss (unchanged) ----------
def quantile_huber_loss(pred, target, tau, delta=1.0):
    error = target - pred
    huber = torch.where(torch.abs(error) < delta, 0.5 * error**2, delta * (torch.abs(error) - 0.5*delta))
    loss = torch.abs(tau - (error < 0).float()) * huber
    return loss.mean()

# ---------- Setup ----------
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
max_degree_of_graph = max(len(list(G.neighbors(n))) for n in G.nodes)
action_dim = max_degree_of_graph
state_dim = 6
#action_dim = 20
batch_size = 32
gamma = 0.99
num_quantiles = 32
learning_rate = 1e-3
target_update_freq = 50      # CHANGED: update more often
episodes = 1000            # CHANGED: increased from 3000 to 10000
max_traffic = max(item["traffic"] for item in iqn_data) if iqn_data else 500.0
env = RoutingEnv(G, max_traffic=max_traffic)
#env = RoutingEnv(G)          # Make sure RoutingEnv has positive reward for done (+10)
iqn_model = IQN(state_dim, action_dim).to(device)
target_model = IQN(state_dim, action_dim).to(device)
target_model.load_state_dict(iqn_model.state_dict())
optimizer = optim.Adam(iqn_model.parameters(), lr=learning_rate)
replay_buffer = ReplayBuffer(100000)

print("🚀 Starting IQN Training (improved settings)...\n")

for episode in range(episodes):
    sample = random.choice(iqn_data)
    src, dst, traffic = sample["src"], sample["dst"], sample["traffic"]
    state = env.reset(src, dst, traffic)
    done = False
    total_reward = 0
    step_count = 0

    while not done:
        #neighbors = get_neighbors(G, state)
        neighbors = sorted(get_neighbors(G, state, dst))
        if not neighbors:
            break

        state_vec = encode_state(state, dst, traffic, G)
        state_tensor = torch.tensor(state_vec, dtype=torch.float32).unsqueeze(0).to(device)

        # ================= DEBUG BLOCK START =================
        # if episode % 50 == 0 and random.random() > 0.9:
        #     with torch.no_grad():
        #         tau = torch.tensor([[0.5]], dtype=torch.float32).to(device)
        #         q_vals = iqn_model(state_tensor, tau).squeeze().cpu().numpy()
        #         q_vals = np.array(q_vals).flatten()
        
        #     print("\nSTATE:", state, "→ DST:", dst)
        #     print("NEIGHBORS:", neighbors)
        #     print("Q_VALUES (raw):", q_vals[:len(neighbors)])
        
        #     best_idx = np.argmax(q_vals[:len(neighbors)])
        #     print("BEST NEIGHBOR:", neighbors[best_idx])
        # ================= DEBUG BLOCK END =================

        # CHANGED: slower epsilon decay (explore longer)
        eps = max(0.01, 1.0 - episode / 2000)   # was /200
        if random.random() < eps:
            action_idx = random.randint(0, len(neighbors)-1)
        else:
            with torch.no_grad():
                tau_median = torch.tensor([[0.5]], dtype=torch.float32).to(device)
                q_vals = iqn_model(state_tensor, tau_median).squeeze().cpu().numpy()
            q_vals = np.array(q_vals).flatten()
            q_vals = q_vals[:len(neighbors)]
            q_vals = np.nan_to_num(q_vals, nan=-np.inf)

            # Always match neighbors length EXACTLY
            if len(q_vals) >= len(neighbors):
                q_vals = q_vals[:len(neighbors)]
            else:
                q_vals = np.pad(q_vals, (0, len(neighbors) - len(q_vals)), constant_values=-np.inf)
            
            # Remove NaNs
            q_vals[np.isnan(q_vals)] = -np.inf
            
            action_idx = int(np.argmax(q_vals))
            action_idx = max(0, min(action_idx, len(neighbors) - 1))
            #action = neighbors[action_idx]

        action = neighbors[action_idx]
        if action not in G[state]:
            print(f"[TRAIN ERROR] Invalid edge: {state} → {action}")
            break
        next_state, reward, done = env.step(action)

        # CHANGED: add a small step penalty to encourage shorter paths
        reward -= 0.05   # penalty per step (makes reaching destination more attractive)

        # Store transition
        next_state_vec = encode_state(next_state, dst, traffic, G) if not done else np.zeros(state_dim)
        replay_buffer.push(state_vec, action_idx, reward, next_state_vec, done)

        state = next_state
        total_reward += reward
        step_count += 1
        if step_count > 50:
            # CHANGED: add a small negative reward for timeout
            total_reward -= 1.0
            break

        # Training step
        if len(replay_buffer) >= batch_size:
            states, actions, rewards, next_states, dones = replay_buffer.sample(batch_size)

            states = torch.tensor(states, dtype=torch.float32).to(device)
            actions = torch.tensor(actions, dtype=torch.long).to(device)
            rewards = torch.tensor(rewards, dtype=torch.float32).to(device)
            next_states = torch.tensor(next_states, dtype=torch.float32).to(device)
            dones = torch.tensor(dones, dtype=torch.float32).to(device)

            tau = torch.rand(batch_size, num_quantiles, device=device)
            tau_next = torch.rand(batch_size, num_quantiles, device=device)

            current_q = iqn_model(states, tau)

            with torch.no_grad():
                next_q = target_model(next_states, tau_next)
                next_q_mean = next_q.mean(dim=1)
                next_action = next_q_mean.argmax(dim=1, keepdim=True)
                target_q = next_q.gather(2, next_action.unsqueeze(1).expand(-1, num_quantiles, -1)).squeeze(2)
                target = rewards.unsqueeze(1) + gamma * (1 - dones.unsqueeze(1)) * target_q

            current_q_taken = current_q.gather(2, actions.unsqueeze(1).unsqueeze(2).expand(-1, num_quantiles, -1)).squeeze(2)
            loss = quantile_huber_loss(current_q_taken.unsqueeze(-1), target.unsqueeze(-1), tau.unsqueeze(-1))

            optimizer.zero_grad()
            loss.backward()
            optimizer.step()

    # Update target network
    if episode % target_update_freq == 0:
        target_model.load_state_dict(iqn_model.state_dict())

    if episode % 100 == 0:   # CHANGED: print every 100 episodes for less clutter
        print(f"Episode {episode}, Total Reward: {total_reward:.2f}, Buffer size: {len(replay_buffer)}")

print("✅ IQN Training finished")
```

```python
import shap
import numpy as np
import matplotlib.pyplot as plt

def explain_tcn_features(model, X_seq, feature_names, src, dst, horizon=3):
    """
    Explain which features matter most for TCN predictions using KernelExplainer.
    
    Args:
        model: trained Keras TCN model (input shape: (batch, window, n_features))
        X_seq: all sequences for this pair, shape (n_samples, window, n_features)
        feature_names: list of feature names (length n_features)
        src, dst: source and destination (for plot title)
        horizon: prediction horizon (number of output steps)
    
    Returns:
        avg_importance: array of shape (n_features,) – mean absolute SHAP per feature
    """
    window_size = X_seq.shape[1]
    n_features = len(feature_names)
    
    # Prediction function for KernelExplainer (flattened input -> output)
    def predict_fn(x_flat):
        # x_flat: (batch, window * n_features)
        x_reshaped = x_flat.reshape(-1, window_size, n_features)
        preds = model.predict(x_reshaped, verbose=0)  # (batch, horizon)
        return preds[:, 0]  # explain first forecast step
    
    # Background (subset of data)
    n_background = min(100, len(X_seq) // 2)
    bg_idx = np.random.choice(len(X_seq), n_background, replace=False)
    background = X_seq[bg_idx]
    background_flat = background.reshape(n_background, -1)
    
    # Test samples (subset)
    n_test = min(50, len(X_seq) // 4)
    test_idx = np.random.choice(len(X_seq), n_test, replace=False)
    test_samples = X_seq[test_idx]
    test_samples_flat = test_samples.reshape(n_test, -1)
    
    # Explain
    explainer = shap.KernelExplainer(predict_fn, background_flat)
    shap_values_flat = explainer.shap_values(test_samples_flat)  # (n_test, window * n_features)
    
    # Reshape to (n_test, window, n_features)
    shap_values = shap_values_flat.reshape(n_test, window_size, n_features)
    
    # Aggregate over time dimension (mean absolute across time and samples)
    # To get a single importance per original feature, we average absolute SHAP over samples and time.
    avg_importance = np.mean(np.abs(shap_values), axis=(0, 1))  # shape (n_features,)
    
    # For summary plot, we need a 2D SHAP matrix (samples × features) after time‑aggregation.
    # We sum SHAP values over time (keeping sign) to preserve directional information.
    shap_time_aggregated = np.sum(shap_values, axis=1)  # (n_test, n_features)
    
    # Also prepare the corresponding test samples aggregated over time (optional, for feature values)
    # For bar plot, only SHAP values are needed; we can pass None for features.
    plt.figure(figsize=(10, 6))
    shap.summary_plot(shap_time_aggregated, features=None, feature_names=feature_names,
                      plot_type="bar", show=False)
    plt.title(f"TCN Feature Importance for {src} → {dst}")
    plt.tight_layout()
    plt.show()
    
    return avg_importance

# ========== USAGE inside your existing loop ==========
# After training 'model' for a pair (src, dst):
feature_names = selected_features  # list of your 15 features
avg_importance = explain_tcn_features(model, X_seq, feature_names, src, dst)

# Print top 5 features
top_indices = np.argsort(avg_importance)[::-1][:5]
print(f"Top 5 features for {src} → {dst}")
for i in top_indices:
    print(f"  {feature_names[i]}: {avg_importance[i]:.4f}")
```
